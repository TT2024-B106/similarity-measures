# Comparación con `Shapely`

```python
import shapely
import json
```

```python
with open('map.geojson') as f:
    gjson = json.load(f)
```

```python
t1 = gjson['features'][0]['geometry']['coordinates']
t2 = gjson['features'][1]['geometry']['coordinates']
l1 = shapely.from_geojson('{"type": "LineString", "coordinates":' + str(t1) + "}")
l2 = shapely.from_geojson('{"type": "LineString", "coordinates":' + str(t2) + "}")

shapely.hausdorff_distance(l1, l2), shapely.frechet_distance(l1, l2)
```

```python
geojson = stmeasures.read_file("../assets/ECATEPEC.json")
t3, t4 = geojson[31], geojson[48]
stmeasures.distance(t3, t4, stmeasures.Algorithms.HAUSDORFF), stmeasures.distance(t3, t4, stmeasures.Algorithms.FRECHET)
```

```python
l3 = shapely.from_geojson('{"type": "LineString", "coordinates":' + str([[lat, lon] for lat, lon in t3]) + "}")
l4 = shapely.from_geojson('{"type": "LineString", "coordinates":' + str([[lat, lon] for lat, lon in t4]) + "}")

shapely.hausdorff_distance(l3, l4), shapely.frechet_distance(l3, l4)
```
