### **Importar biblioteca**
Para utilizar esta biblioteca, primero se debe importar, y luego se crean referencias a los módulos principales para facilitar la lectura y el uso del código. A continuación se muestra el código de ejemplo:

```python
import stmeasures.read as read
import stmeasures.visualize as visualize
import geojsonio
import json
```

### **Cargar el archivo GeoJSON:** 
Utilizamos el módulo read de la biblioteca stmeasures para cargar un archivo en formato GeoJSON, en este caso, ECATEPEC.json. Esto convierte el archivo en un objeto manipulable dentro del código.

```python
geojson_obj = read.file("ECATEPEC.json")
```

### **Visualizar una trayectoria en geojson.io:** 
Usamos el módulo visualize para convertir una trayectoria específica del archivo (geojson_obj.trajectories[100]) en un formato compatible con geojson.io. La trayectoria se convierte en JSON y se envía al sitio web geojson.io para su visualización interactiva.

```python
geojsonio.display(json.dumps(visualize.geojsonio_trajectory(geojson_obj.trajectories[100])))
```
