## Pruebas con anteriores trayectorias

Se realizan las pruebas _in vitro_ hechas anteriormente ahora con estos 2 algoritmos

### Primeras trayectorias similares

```python
ecatepec_similar1 = [31, 48]
trajectories1 = utils.get_trajectories(g_ecatepec, ecatepec_similar1)
stmeasures.distance(*trajectories1, stmeasures.Algorithms.HAUSDORFF), stmeasures.distance(*trajectories1, stmeasures.Algorithms.FRECHET)
```

```python
trajectories1_s = utils.convert_to_same_size(*trajectories1)
stmeasures.distance(*trajectories1_s, stmeasures.Algorithms.HAUSDORFF), stmeasures.distance(*trajectories1_s, stmeasures.Algorithms.FRECHET)
```
