## Identificando trayectorias útiles
Las pruebas que se buscan hacer sobre todos los algoritmos son con trayectorias:

1. Muy similares
2. Muy distintas
3. Aletorias

Por lo que primeramente necesitamos identificar estas trayectorias y para esto necesitarémos utilizar nuestro paquete `stmeasures` y `geojsonio`.

```python
import stmeasures
import geojsonio
```

Se utilizará el geojson `ECATEPEC.json` para encontrar estas trayectorias.

```python
geojson = stmeasures.read_file("../assets/ECATEPEC.json")
print(f"GeoJSON con {len(geojson[:])} trayectorias")
```

Definimos una función que nos puede ir ayudando ir visualizando estas trayectorias:

```python
def see_each_trajectory(geojson, start, end):
    for i in range(start, end):
        _ = geojsonio.display(stmeasures.geojsonio_contents(trajectory=geojson[i], indexes=[i]))
```

Y de esta forma podemos ir viendo trayectorias dentro de nuestro objeto `geojson` e ir guardando algunos índices que nos puedan ser útiles.

```python
_iter = 5
_range = 5
start = _range * (_iter - 1)
end = _range * _iter

see_each_trajectory(geojson, start, end)
```

```python
useful_trajectories = [8, 13, ]
# 8 es una línea reacta, quizás podríamos desplazarla, para tener dos lineas "paralelas" y ver como se van comportando nuestros algoritmos
# 8 y 13 son trayectorias algo lejanas
# 
```

```python
other_trajectories = [10, 16, 17, 20, ]
# 10 tiene una forma rara, pero puede servir
# 16 me gusta porque es como de un carro
# 17 puede ser útil, solo que tiene muchos puntos en un lugar
# 18 es como de otro carro
# 20, 21 puede servir (tienen un poco forma rara)
# 
```

```python
car_trajectories = [16, 18, 19, 22, ]
```

Ahora para poder ir viendo las trayectorias que vamos creyendo pueden ser útiles lo podemos hacer con la siguiente functión:

```python
def get_trajectories(geojson, indexes):
    return [geojson[i] for i in indexes]

def see_trajectories(geojson, indexes):
    _ = geojsonio.display(
        stmeasures.geojsonio_contents(trajectories=get_trajectories(geojson, indexes), indexes=indexes)
    )
```

```python
see_trajectories(geojson, useful_trajectories)
```

```python
see_trajectories(geojson, car_trajectories)
```
