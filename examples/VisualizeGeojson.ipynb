## Este módulo proporciona una función para leer archivos GeoJSON y devolver un objeto GeoJSON.

+ **Función file:** Esta función lee el archivo GeoJSON desde la ruta especificada y devuelve un objeto GeoJSON.

### stmeasures/read.py

````python
def file(filepath: str) -> GeoJSON:
    with open(filepath, 'r') as f:
        data = json.load(f)
    return GeoJSON(data)
````

## Este módulo proporciona funciones para visualizar trayectorias individuales o múltiples utilizando geojsonio.

+ **Función swap_coordinates:** Convierte las coordenadas de formato [latitud, longitud] a [longitud, latitud].

+ **Función trajectory:** Visualiza una sola trayectoria en geojsonio.

+ **Función visualize_trajectory_by_index:** Visualiza una trayectoria específica basada en el índice proporcionado.

+ **Función trajectories:** Visualiza todas las trayectorias en el archivo GeoJSON, aunque geojson.io no lo permite

### stmeasures/visualize.py

````python
def swap_coordinates(coords):
    return [[lon, lat] for lat, lon in coords]

def trajectory(trajectory):
    geojson = {
        "type": "LineString",
        "coordinates": trajectory['coordinates']
    }
    geojsonio.display(json.dumps(geojson))

def visualize_trajectory_by_index(geojson_obj, index):
    if 0 <= index < len(geojson_obj.trajectories):
        trajectory(geojson_obj.trajectories[index])
    else:
        print(f"Invalid index: {index}. Please choose an index between 0 and {len(geojson_obj.trajectories) - 1}.")

def trajectories(geojson_obj):
    geojson = {
        "type": "FeatureCollection",
        "features": []
    }
    for trajectory in geojson_obj.trajectories:
        geojson['features'].append({
            "type": "Feature",
            "geometry": {
                "type": "LineString",
                "coordinates": trajectory['coordinates']
            },
            "properties": {
                "id": trajectory['id'],
                "timestamp": trajectory['timestamp']
            }
        })
    geojsonio.display(json.dumps(geojson))
````
